# Processing the files in sequential order.

```python
from docling.document_converter import DocumentConverter
import os

# Configuration
current_dir = os.getcwd()
output_dir = os.path.join(current_dir, "docling_output_md")
os.makedirs(output_dir, exist_ok=True)

# Specify the relative path to the PDF folder from the current directory
pdf_folder_relative_path = "pdfs_20_2"  # Change this to your PDF folder name
pdf_dir = os.path.join(current_dir, pdf_folder_relative_path)

# Processing parameters
OFFSET = 0     # Number of PDFs to skip (start from this index)
LIMIT = 273       # Maximum number of PDFs to process (0 = all)

# Get sorted list of PDF files from the specified folder
pdf_files = [f for f in os.listdir(pdf_dir) if f.lower().endswith(".pdf")]
pdf_files.sort()  # Sort alphabetically

# Apply offset and limit
if LIMIT > 0:
    pdf_files = pdf_files[OFFSET:OFFSET+LIMIT]
else:
    pdf_files = pdf_files[OFFSET:]  # Process all remaining after offset

converter = DocumentConverter()
processed_count = 0

print(f"🚀 Starting conversion: Processing {len(pdf_files)} PDFs from {pdf_dir}")
print(f"📁 PDF folder: {pdf_dir}")
print(f"📄 Output folder: {output_dir}")
print(f"⚙️  Settings: offset={OFFSET}, limit={LIMIT}")
print("-" * 60)

for filename in pdf_files:
    source = os.path.join(pdf_dir, filename)  # Use pdf_dir instead of current_dir
    output_filename = os.path.splitext(filename)[0] + ".md"
    output_path = os.path.join(output_dir, output_filename)

    try:
        # Convert PDF to Markdown
        doc = converter.convert(source).document
        markdown_content = doc.export_to_markdown()

        # Save to file
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(markdown_content)

        processed_count += 1
        print(f"✅ [{processed_count}/{len(pdf_files)}] Converted: {filename}")

    except Exception as e:
        print(f"❌ Failed to convert {filename}: {str(e)}")

print("-" * 60)
print(f"🎉 Conversion complete! Processed {processed_count} files.")
print(f"📁 Output saved in: {output_dir}")
```

# Processing multiple pdf files simultaneously.

```python
# This is not working completly because its crachsing when launched over super memory consuption.
from docling.document_converter import DocumentConverter
import os
import concurrent.futures
import traceback

# Configuration
current_dir = os.getcwd()
output_dir = os.path.join(current_dir, "docling_output_md_concurrent")
os.makedirs(output_dir, exist_ok=True)

# Specify the relative path to the PDF folder  from the current directory
pdf_folder_relative_path = "pdfs_20_2"  # Change this to your PDF folder name
pdf_dir = os.path.join(current_dir, pdf_folder_relative_path)

# Processing parameters
OFFSET = 0  # Number of PDFs to skip (start from this index)
LIMIT = 273        # Maximum number of PDFs to process (0 = all)
MAX_WORKERS = 2  # adjust number of threads here

# Get sorted list of PDF files from the specified folder
if not os.path.isdir(pdf_dir):
    raise FileNotFoundError(f"PDF directory does not exist: {pdf_dir}")

pdf_files = [f for f in os.listdir(pdf_dir) if f.lower().endswith(".pdf")]
pdf_files.sort()  # Sort alphabetically

# Apply offset and limit
if LIMIT > 0:
    pdf_files = pdf_files[OFFSET:OFFSET+LIMIT]
else:
    pdf_files = pdf_files[OFFSET:]  # Process all remaining after offset

print(f"🚀 Starting conversion: Processing {len(pdf_files)} PDFs from {pdf_dir}")
print(f"📁 PDF folder: {pdf_dir}")
print(f"📄 Output folder: {output_dir}")
print(f"⚙️  Settings: offset={OFFSET}, limit={LIMIT}, max_workers={MAX_WORKERS}")
print("-" * 60)

def process_pdf(filename):
    """Worker: create own DocumentConverter instance to avoid shared-state issues."""
    source = os.path.join(pdf_dir, filename)
    output_filename = os.path.splitext(filename)[0] + ".md"
    output_path = os.path.join(output_dir, output_filename)
    try:
        converter = DocumentConverter()
        doc = converter.convert(source).document
        markdown_content = doc.export_to_markdown()
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(markdown_content)
        return (filename, None)
    except Exception as e:
        return (filename, traceback.format_exc())

processed_count = 0
failed = []

with concurrent.futures.ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
    futures = {executor.submit(process_pdf, fn): fn for fn in pdf_files}
    for fut in concurrent.futures.as_completed(futures):
        filename = futures[fut]
        try:
            fn, err = fut.result()
            processed_count += 1
            if err is None:
                print(f"✅ [{processed_count}/{len(pdf_files)}] Converted: {fn}")
            else:
                failed.append((fn, err))
                print(f"❌ [{processed_count}/{len(pdf_files)}] Failed: {fn}\n{err}")
        except Exception as e:
            failed.append((filename, traceback.format_exc()))
            processed_count += 1
            print(f"❌ [{processed_count}/{len(pdf_files)}] Failed (executor error): {filename}\n{traceback.format_exc()}")

print("-" * 60)
print(f"🎉 Conversion complete! Processed {processed_count} files. Success: {processed_count - len(failed)}  Failed: {len(failed)}")
print(f"📁 Output saved in: {output_dir}")
# ...existing code...
```
